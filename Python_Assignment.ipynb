```python
#1.FizzBuzz
for i in range(1,51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
#2.Leap_Year
def is_leap_year(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False


year = 2024
print(is_leap_year(year))
```

```python
#3.Identity vs Equality
L1 = [1,2,3]
L2 = [1,2,3]

print(L1 == L2)
print(L1 is L2)

L2 = L1

print(L1 == L2)
print(L1 is L2)
```

```python
#4.Bitwise Swap
a = 10
b = 20

a = a ^ b
b = a ^ b
a = a ^ b

print(a)
print(b)
```

```python
#5.Prime Number Finder
count = 0
number = 2

while count < 10:
    prime = True
    i = 2

    while i < number:
        if number % i == 0:
            prime = False
            break
        i += 1

    if prime:
        print(number)
        count += 1

    number += 1
```

```python
#6.Loop Breaker
for i in range(1,21):
    if i == 18:
        break

    if i == 13:
        continue

    if i % 4 == 0:
        pass
    else:
        print(i)
```

```python
#7.Nested conditions
marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Fail")
```

```python
#8.reverse an integer
number = 12345
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

print(reverse)
```

```python
#9. runnerup score
scores = [10,20,40,30,40,50]

highest = max(scores)

second_highest = None

for score in scores:
    if score != highest:
        if second_highest is None or score > second_highest:
            second_highest = score

print(second_highest)
```

```python
#10. duplicate removal
numbers = [1,2,2,3,4,4,5]

unique_values = list(set(numbers))

print(unique_values)
```

```python
#11. list intersection
list1 = [1,2,3,4,5]
list2 = [3,4,5,6,7]

common = []

for item in list1:
    if item in list2:
        common.append(item)

print(common)
```

```python
#12. tuple Immutability
numbers = (10,20,30)

try:
    numbers[1] = 50
except TypeError:
    print("Tuple cannot be changed")
```

```python
#13.Character frequency counter
text = "hello"

frequency = {}

for character in text:
    if character in frequency:
        frequency[character] += 1
    else:
        frequency[character] = 1

print(frequency)
```

```python
#14.list comprehension
squares = [i*i for i in range(1,21) if i % 2 == 0]

print(squares)
```

```python
#15.dictionaey comprehension
data = {'a':1, 'b':2, 'c':3, 'd':4}

result = {key:value for key,value in data.items() if value > 2}

print(result)
```
